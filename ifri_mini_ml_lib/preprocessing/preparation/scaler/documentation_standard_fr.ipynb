## Description
Le standardScaler est utilisé pour standariser les données c'est a dire les  mettre a la meme échelle. Il calcule la moyenne et l"écart-type de chaque colonne puis transforme les valeurs pour que chaque colonne ait une moyenne de 0 et un écart-type de 1.

### Formule
***X' = (X - moyenne) / écart-type.***

La mise en place et le fonctionnement du standardScaler se base de six (06) méthode dont: fit(); transform(); fit_transform(); inverse_transform(); convert_to_array(); _convert_to_dataframe().

## Méthode

### fit()
La méthode fit() a pour role de calculer et de mémoriser la moyenne et l'écart-type de chaque colonne. Elle prend en entrée un tableau de données X, calcul la moyenne avec np.mean() et l'écart-type avec np.std() pour chaque colonne puis il les mémorise dans self.mean() et self.std()afin de retourner le scaler.

###  transform ()
La méthode transform() a pour role de standariser les données en appliquant la formule Z-score. Elle prend en entrée un tableauX, vérifie que fit() a bien été appelé, applique la formule X' = (X - moyenne) / écart-type et retourne les données standarisées dans le meme format que l'entrée.

### fit_transform()
La méthode fit_transform est une union des deux première méthode.Elle appelle fit(X) pour calculer la moyenne et l'écart-type, puis transform(X) standarise et retourne les données standarisées.

### inverse_transform()
La méthode inverse_transform() a pour role de remettre les données dans leurs échelle d'origine. Elle prend des données standarisées, Vérifie que fit() a été appelé, applique la formule inverse (X* écart-type) + moyenne et retourne les données originales.

### convert_to_array()
La méthode convert_to_array() a pour role de convertir les données d'entrée en tableau numpy. Elle prend en entrée des données X et les convertit dans le but de permettre les calculs. Si X est un dataframe, elle mémorise les noms de colonne afin de retourner les valeurs. Dans d'autres cas, elle convertit directement en tableau numpy.

### convert_to_dataframe()
La méthode convert_to_dataframe() a pour role de remettre les données dans leurs format d'origine. Elle prend en entrée le tableau numpy transformé afin de le reconvertit vers le format initial. Dans le cas ou les données originales était un dataframe, elle mémorise les noms de colonne et recrée un dataframe. Sinon, elle retourne le tableau numpay sans rien changé.

```python

import numpy as np
import pandas as pd
import time
from sklearn.preprocessing import StandardScaler as SklearnStandardScaler

### Chargement du dataset
df = pd.read_csv('base3.csv')

### Sélection de 3 colonnes numériques
data = df[['Age', 'EstimatedSalary', 'Purchased']]

print("Dataset original :")
print(data.head(10))

### Notre implémentation
debut = time.time()
notre_scaler = StandardScaler()
notre_resultat = notre_scaler.fit_transform(data)
notre_temps = time.time() - debut

### Scikit-learn
debut = time.time()
sklearn_scaler = SklearnStandardScaler()
sklearn_resultat = sklearn_scaler.fit_transform(data)
sklearn_temps = time.time() - debut

print("Notre StandardScaler :")
print(notre_resultat)
print(f"Temps : {notre_temps:.6f} secondes")

print("\nScikit-learn StandardScaler :")
print(sklearn_resultat)
print(f"Temps : {sklearn_temps:.6f} secondes")

print("\nRésultats identiques :", np.allclose(notre_resultat, sklearn_resultat))
```

## Noté Bien

### MinMaxScaler
- Ramène les valeurs entre 0 et 1
- Utilise la formule (X - min) / (max - min)
- Très sensible aux valeurs extrêmes
- Garde des bornes fixes [0, 1]
- À utiliser quand on connaît les bornes des données

### StandardScaler
- Centre les valeurs autour de 0
- Utilise la formule (X - moyenne) / écart-type
- Moins sensible aux valeurs extrêmes
- Pas de bornes fixes
- À utiliser quand les données suivent une distribution normale
