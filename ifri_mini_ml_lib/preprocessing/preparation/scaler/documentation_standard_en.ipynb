### Description
- The StandardScaler is used to **standardize data** by bringing it to the same scale
- It calculates the **mean** and **standard deviation** of each column
- It transforms values so that each column has :
  - A mean of **0**
  - A standard deviation of **1**

### Formula
- **X' = (X - mean) / standard deviation**

### Methods
The StandardScaler is based on **6 methods** :

#### 1. fit()
- **Role** : Calculates and stores the mean and standard deviation
- **Input** : Data array X
- **Process** :
  - Calculates mean using `np.mean()`
  - Calculates standard deviation using `np.std()`
  - Stores them in `self.mean_` and `self.std_`
- **Output** : Returns the scaler object

#### 2. transform()
- **Role** : Standardizes the data using the Z-score formula
- **Input** : Data array X
- **Process** :
  - Verifies that `fit()` has been called
  - Applies formula **X' = (X - mean) / standard deviation**
- **Output** : Standardized data in the same format as input

#### 3. fit_transform()
- **Role** : Combines `fit()` and `transform()` in one step
- **Process** :
  - Calls `fit(X)` to calculate mean and standard deviation
  - Calls `transform(X)` to standardize
- **Output** : Standardized data


#### 4. inverse_transform()
- **Role** : Restores data to its original scale
- **Input** : Standardized data
- **Process** :
  - Verifies that `fit()` has been called
  - Applies inverse formula **(X * standard deviation) + mean**
- **Output** : Original data


#### 5. _convert_to_array()
- **Role** : Converts input data into a NumPy array
- **Input** : Data X
- **Process** :
  - If X is a DataFrame → stores column names and returns values
  - Otherwise → converts directly to NumPy array
- **Output** : NumPy array


#### 6. _convert_to_dataframe()
- **Role** : Restores data to its original format
- **Input** : Transformed NumPy array
- **Process** :
  - If original data was a DataFrame → recreates DataFrame with column names
  - Otherwise → returns NumPy array unchanged
- **Output** : Data in original format

```python
import numpy as np
import pandas as pd
import time
from sklearn.preprocessing import StandardScaler as SklearnStandardScaler

### Loading the dataset
df = pd.read_csv('base3.csv')

### Selecting 3 numerical columns
data = df[['Age', 'EstimatedSalary', 'Purchased']]

print("Original dataset:")
print(data.head(10))

### our implementation
start = time.time()
our_scaler = StandardScaler()
our_result = our_scaler.fit_transform(data)
our_time = time.time() - start

### Scikit-learn
start = time.time()
sklearn_scaler = SklearnStandardScaler()
sklearn_result = sklearn_scaler.fit_transform(data)
sklearn_time = time.time() - start

print("our StandardScaler :")
print(our_result)
print(f"Time : {our_time:.6f} seconds")

print("\nScikit-learn StandardScaler :")
print(sklearn_result)
print(f"Time : {sklearn_time:.6f} seconds")

print("\nIdentical results :", np.allclose(our_result, sklearn_result))
```

### 3. MinMaxScaler vs StandardScaler

**MinMaxScaler :**
- Scales values between **0 and 1**
- Formula : **(X - min) / (max - min)**
- Very sensitive to outliers
- Maintains fixed bounds **[0, 1]**
- Use when **data bounds are known**

**StandardScaler :**
- Centers values around **0**
- Formula : **(X - mean) / standard deviation**
- Less sensitive to outliers
- No fixed bounds
- Use when **data follows a normal distribution**
