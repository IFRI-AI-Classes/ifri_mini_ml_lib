# MinMaxScaler

## Introduction

Feature scaling is a crucial preprocessing step in machine learning pipelines. 
When features have different scales, algorithms that rely on distances or 
gradients can be heavily influenced by features with larger magnitudes.

The **MinMaxScaler** is a normalization technique that transforms each feature 
by mapping its values to a fixed range, typically **[0, 1]**. This ensures 
that all features contribute equally to the model, regardless of their 
original scale.

## 1. Key Concepts

Feature scaling is a preprocessing technique used in machine learning 
to bring all features to the same scale. Without scaling, features with 
large values can dominate those with smaller values, leading to biased 
or inaccurate model predictions.

For example, consider a dataset with two features:
- **Age**: values between 18 and 60
- **Salary**: values between 15,000 and 150,000

Without scaling, the salary feature would dominate the age feature 
simply because of its larger magnitude. After applying MinMaxScaler, 
both features are mapped to **[0, 1]**, giving them equal importance.

MinMaxScaler is particularly useful when:
- The data does **not** follow a normal distribution
- The algorithm requires bounded input values (e.g. neural networks)
- You want to preserve the original distribution of the data

## 2. Mathematical Formula

The MinMaxScaler applies the following formula to each value in a feature :

$$X' = \frac{X - X_{min}}{X_{max} - X_{min}}$$

Where :
- $X$ : the original value
- $X_{min}$ : the minimum value of the feature
- $X_{max}$ : the maximum value of the feature
- $X'$ : the normalized value (between 0 and 1)

**Step-by-step example :**

Consider the feature **Age** : [18, 25, 35, 45, 60]
- $X_{min}$ = 18
- $X_{max}$ = 60

| Original value | Calculation | Normalized value |
|---|---|---|
| 18 | (18 - 18) / (60 - 18) | 0.000 |
| 25 | (25 - 18) / (60 - 18) | 0.167 |
| 35 | (35 - 18) / (60 - 18) | 0.405 |
| 45 | (45 - 18) / (60 - 18) | 0.643 |
| 60 | (60 - 18) / (60 - 18) | 1.000 |

## 3. Pseudo-algorithm

**Input :** $X$ : dataset of $n$ samples and $p$ features

---

**fit($X$)**
- Compute $X_{min}$ and $X_{max}$ for each feature
- Store them for later use

**transform($X$)**
- For each value $x$ : $X' = \frac{x - X_{min}}{X_{max} - X_{min}}$
- Return $X'$ in $[0, 1]$

**fit_transform($X$)**
- Call fit($X$) then transform($X$)
- Return normalized data

**inverse_transform($X'$)**
- For each value $x'$ : $X = x' \times (X_{max} - X_{min}) + X_{min}$
- Return original data

**_convert_to_array($X$)**
- Convert DataFrame/Series/array → NumPy array

**_convert_back($X'$)**
- Convert NumPy array → original format (DataFrame/Series/array)

## 6. Real-life Applications

Because of its simplicity and effectiveness, MinMaxScaler is widely 
used in various domains :

- **Neural Networks** : Neural networks require input values between 
  0 and 1 for faster and more stable convergence during training.

- **Image Processing** : Pixel values (0-255) are normalized to [0, 1] 
  before feeding images into machine learning models.

- **Finance** : Stock prices and financial indicators with very different 
  scales are normalized before being used in predictive models.

- **Healthcare** : Patient data such as age, weight, and blood pressure 
  are normalized to give equal importance to each feature.

## 7. Limitations and Challenges

- **Sensitive to outliers** : If a feature contains extreme values, 
  they will compress all other values into a very small range, 
  reducing the model's ability to distinguish between them.

- **Fixed range** : The output is always bounded to [0, 1]. 
  If new data contains values outside the training range, 
  the scaler will produce values outside [0, 1].

- **Division by zero** : If all values in a feature are identical, 
  $X_{max} - X_{min} = 0$, which causes a division by zero error.

- **Does not handle pd.Series** : Unlike StandardScaler, 
  our implementation does not support pandas Series as input.
