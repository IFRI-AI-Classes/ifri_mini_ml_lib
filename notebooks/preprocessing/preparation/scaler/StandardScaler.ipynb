# StandardScaler

## Introduction

Feature scaling is a crucial preprocessing step in machine learning 
pipelines. When features have different scales, algorithms that rely 
on distances or gradients can be heavily influenced by features with 
larger magnitudes.

The **StandardScaler** is a standardization technique that transforms 
each feature by centering it around **0** and scaling it to a standard 
deviation of **1**. This ensures that all features contribute equally 
to the model, regardless of their original scale.

## 1. Key Concepts

Feature standardization is a preprocessing technique used in machine 
learning to bring all features to the same scale. Unlike normalization 
which maps values to [0, 1], standardization centers the data around 
**0** with a standard deviation of **1**.

For example, consider a dataset with two features:
- **Age**: values between 18 and 60
- **Salary**: values between 15,000 and 150,000

Without scaling, the salary feature would dominate the age feature 
simply because of its larger magnitude. After applying StandardScaler, 
both features are centered around 0 with a standard deviation of 1, 
giving them equal importance.

StandardScaler is particularly useful when:
- The data follows a **normal distribution**
- The algorithm is sensitive to the **variance** of features 
  (e.g. PCA, SVM, Linear Regression)
- There are **outliers** in the data — StandardScaler is less 
  sensitive to outliers than MinMaxScaler

## 2. Mathematical Formula

The StandardScaler applies the following formula to each value in a feature :

$$X' = \frac{X - \mu}{\sigma}$$

Where :
- $X$ : the original value
- $\mu$ : the mean of the feature
- $\sigma$ : the standard deviation of the feature
- $X'$ : the standardized value

**Step-by-step example :**

Consider the feature **Age** : [18, 25, 35, 45, 60]
- $\mu$ = 36.6
- $\sigma$ = 15.94

| Original value | Calculation | Standardized value |
|---|---|---|
| 18 | (18 - 36.6) / 15.94 | -1.167 |
| 25 | (25 - 36.6) / 15.94 | -0.728 |
| 35 | (35 - 36.6) / 15.94 | -0.100 |
| 45 | (45 - 36.6) / 15.94 | +0.527 |
| 60 | (60 - 36.6) / 15.94 | +1.468 |

## 3. Pseudo-algorithm

**Input :** $X$ : dataset of $n$ samples and $p$ features

---

**fit($X$)**
- Compute $\mu$ (mean) and $\sigma$ (standard deviation) for each feature
- Store them for later use

**transform($X$)**
- For each value $x$ : $X' = \frac{x - \mu}{\sigma}$
- Return standardized data centered around 0

**fit_transform($X$)**
- Call fit($X$) then transform($X$)
- Return standardized data

**inverse_transform($X'$)**
- For each value $x'$ : $X = x' \times \sigma + \mu$
- Return original data

**_convert_to_array($X$)**
- Convert DataFrame/Series/array → NumPy array

**_convert_to_dataframe($X'$)**
- Convert NumPy array → original format (DataFrame/Series/array)

## 4. Implementation

For this implementation, we use the **base3.csv** dataset which contains 
information about users and their purchasing behavior.

The numerical features used are : **Age**, **EstimatedSalary** and **Purchased**.

```python
import pandas as pd
import numpy as np
from ifri_mini_ml_lib.preprocessing.preparation.scaler.standard_scaler import StandardScaler

# Load dataset
df = pd.read_csv('base3.csv')
data = df[['Age', 'EstimatedSalary', 'Purchased']]

# Illustrative example
scaler = StandardScaler()
result = scaler.fit_transform(data)

print("Original data (first 5 rows):")
print(data.head())
print("\nStandardized data (first 5 rows):")
print(result.head())
```

## 6. Real-life Applications

Because of its effectiveness with normally distributed data, 
StandardScaler is widely used in various domains :

- **Natural Language Processing (NLP)** : Text features such as 
  word frequencies are standardized before being fed into 
  classification models.

- **Medical Diagnosis** : Patient data such as blood pressure, 
  cholesterol levels and age are standardized to give equal 
  importance to each feature in diagnostic models.

- **Finance** : Financial indicators such as stock returns and 
  volatility are standardized before being used in risk assessment 
  and prediction models.

- **Computer Vision** : Image pixel values are standardized 
  before training deep learning models to improve convergence speed.

## 7. Limitations and Challenges

- **Assumes normal distribution** : StandardScaler works best when 
  the data follows a normal distribution. If the data is heavily 
  skewed, standardization may not be the best choice.

- **Division by zero** : If all values in a feature are identical, 
  $\sigma = 0$, which causes a division by zero error.

- **Does not bound output** : Unlike MinMaxScaler, StandardScaler 
  does not produce values in a fixed range. New data with extreme 
  values can produce very large or very small standardized values.

- **Does not handle pd.Series** : Our implementation does not 
  support pandas Series as input, unlike MinMaxScaler.
