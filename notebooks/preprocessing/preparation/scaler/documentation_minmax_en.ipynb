## Description
- The MinMaxScaler is used to **normalize data** by mapping each value
  to the range **[0, 1]**
- Formula used : **(X - min) / (max - min)**
- Prevents a feature with large values from **dominating** the others
  during model training

## Formula
- **X' = (X - min) / (max - min)**

---

## Methods
The MinMaxScaler is based on **7 methods** :

#### 1. fit()
- **Role** : Calculates and stores the min and max values
- **Input** : Data array X
- **Process** :
  - Calculates and stores min and max of each column
  - Must be called **before** `transform()`
- **Output** : Returns the scaler object for method chaining

#### 2. transform()
- **Role** : Normalizes data into the interval **[0, 1]**
- **Input** : Data array X
- **Process** :
  - Verifies that `fit()` has been called (raises `ValueError` otherwise)
  - Applies formula **(X - min) / (max - min)**
- **Output** : Transformed data in the same format as input

#### 3. fit_transform()
- **Role** : Combines `fit()` and `transform()` in one step
- **Process** :
  - Calls `fit(X)` to calculate min and max
  - Calls `transform(X)` to normalize
- **Output** : Normalized data in **[0, 1]**

#### 4. inverse_transform()
- **Role** : Reconstructs original data from normalized data
- **Input** : Normalized data array
- **Process** :
  - Verifies that `fit()` has been called (raises `ValueError` otherwise)
  - Applies inverse formula **min + X_scaled * (max - min)**
- **Output** : Data restored to original scale

#### 5. _convert_to_array()
- **Role** : Converts input data into a NumPy array
- **Input** : Data X
- **Process** :
  - If X is a **DataFrame** → stores column names, returns values as array
  - If X is a **Series** → stores column name, returns values
  - Otherwise → converts directly to NumPy array
- **Output** : Always returns a **NumPy array**

#### 6. _convert_back()
- **Role** : Converts NumPy array back to original format
- **Input** : Transformed NumPy array + original data as reference
- **Process** :
  - If original was a **DataFrame** → recreates DataFrame with stored column names
  - If original was a **Series** → recreates Series with stored column name
  - Otherwise → returns NumPy array as-is
- **Output** : Data in original format

### 3. Comparison : MinMaxScaler vs StandardScaler

**MinMaxScaler :**
- Scales values between **0 and 1**
- Formula : **(X - min) / (max - min)**
- Very sensitive to outliers
- Fixed bounds **[0, 1]**
- Use when **data bounds are known**

**StandardScaler :**
- Centers values around **0**
- Formula : **(X - mean) / standard deviation**
- Less sensitive to outliers
- No fixed bounds
- Use when **data follows a normal distribution**
