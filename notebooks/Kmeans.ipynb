# K-Means Algorithm





K-Means is an unsupervised machine learning algorithm used for clustering.

The objective of K-Means is to partition data into K distinct clusters such that points inside the same cluster are similar to each other.

The algorithm minimizes the distance between data points and their assigned cluster centroids.

# 1.Key Concepts





### Cluster
A cluster is a group of similar data points.

### Centroid
A centroid is the center of a cluster.

### Iteration
K-Means works iteratively by updating centroids until convergence.

### Convergence
The algorithm stops when centroid positions no longer significantly change.

# 2-Distance Metrics




K-Means commonly uses Euclidean distance to measure similarity between points.
$$
d(x,y)=\sqrt{\sum_{i=1}^{n}(x_i-y_i)^2}




Where:
- x and y are data points
- n is the number of features

Smaller distances indicate higher similarity between points.

# 3. Pseudo Algorithm

The K-Means algorithm follows these steps:

1. Choose the number of clusters K
2. Initialize K centroids
3. Assign each point to the nearest centroid
4. Update centroids using cluster means
5. Repeat until convergence

## Objective Function

$$
J=\sum_{i=1}^{k}\sum_{x_j\in C_i}||x_j-\mu_i||^2

Where:

- \( J \) : Objective function minimized by K-Means
- \( k \) : Number of clusters
- \( Ci \) : Set of points belonging to cluster \( i \)
- \( xj \) : A data point
- \( µi \) : Centroid of cluster \( i \)
- \( ||xj - µi||^2 \) : Squared Euclidean distance between a point and its centroid

# 7. Limitations and Challenges

Although K-Means is simple and efficient, it also presents several limitations and challenges.

## Sensitive to Outliers

K-Means uses the mean position of points to compute centroids.
Because of this, extreme values (outliers) can significantly affect centroid positions.

As a result:
- clusters may become distorted,
- centroid locations may shift incorrectly,
- clustering quality may decrease.

---

## Requires Selecting the Number of Clusters in Advance

The user must specify the number of clusters \( K \) before training the model.

Choosing an inappropriate value for \( K \) can lead to:
- over-clustering,
- under-clustering,
- poor data separation.

Methods such as the Elbow Method are often used to estimate an appropriate number of clusters.

---

## Performs Poorly on Non-Spherical Clusters

K-Means assumes that clusters are approximately spherical and equally distributed.

When clusters have:
- irregular shapes,
- varying densities,
- overlapping regions,

the algorithm may incorrectly assign points to clusters.

Algorithms such as DBSCAN are often more suitable for such datasets.

---

## Sensitive to Centroid Initialization

The initial positions of centroids strongly influence the final clustering result.

Poor initialization may cause:
- slow convergence,
- unstable clustering,
- suboptimal solutions.

The K-Means++ initialization strategy helps reduce this problem by selecting better initial centroids.

---

## May Converge to Local Minima

K-Means does not always find the global optimal solution.

Depending on initialization, the algorithm may converge to a local minimum instead of the best clustering configuration.

For this reason, K-Means is often executed multiple times with different random initializations.

---

## Difficulty Handling Clusters with Different Sizes

K-Means performs best when clusters have similar sizes and densities.

If one cluster is much larger or denser than another, the algorithm may produce inaccurate boundaries between groups.

---

## Not Suitable for Categorical Data

K-Means relies on Euclidean distance, which is designed for numerical data.

Therefore, the algorithm is generally not appropriate for:
- text categories,
- labels,
- non-numerical attributes.
