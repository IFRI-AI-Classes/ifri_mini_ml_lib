# [▢]() | The [`Perceptron`](https://en.wikipedia.org/wiki/Perceptron)

The perceptron is a [supervised learning]() algorithm for binary classifiers (i.e., separating two classes). It was invented in 1957 by [Frank Rosenblatt]() at the Cornell University Aeronautics Laboratory. It is a [formal neuron](https://en.wikipedia.org/wiki/Artificial_neuron) equipped with a learning rule that automatically determines [synaptic weights](https://fr.wikipedia.org/wiki/Neurone_formel#Formulation_math%C3%A9matique) to separate a supervised learning problem. If the problem is linearly separable, a theorem ensures that the perceptron rule allows finding a separation between the two classes.

## [▻]() | Definition

The perceptron is the simplest type of direct-action neural network ([***Feedforward** neural network*](https://en.wikipedia.org/wiki/Feedforward_neural_network)). In its initial version, the perceptron is single-layered and has only one output ([Boolean]()) to which all Boolean inputs are connected. More generally, the inputs can be real numbers, and it then [linear classifier](https://en.wikipedia.org/wiki/Linear_classifier) for real vector data.

## [▻]() $M_1^*$ | Perceptron &nbsp; [$"Activation$ $Function"$](https://en.wikipedia.org/wiki/Activation_function)

The outputs of the `Perceptron` result from the application of the [`Heaviside function`](https://en.wikipedia.org/wiki/Heaviside_step_function) (threshold function), the only activation function to train a `single-layer perceptron` $--$ unlike the `multi-layer perceptron` (**[`MLP`](https://en.wikipedia.org/wiki/Multilayer_perceptron)**) which offers several options (at least one for each layer).

where 
- $X = \{{x_1, x_2, ..., x_n}\}$ | $X \in \mathbb{R^{n\times p}}$ et $x_i \in \mathbb{R^p}$, is the set of input data (a vector of vectors, an n-dimensional array)
- $w = \{{w_1, w_2, ..., w_n}\}$ | $w \in \mathbb{R^n}$, each corresponding to an entry ($w_i \rightarrow x_i$ $\Longrightarrow$ $\hat{y} =\sum_{i=0}^{n}w_i \times x_i + b$) and indicating its importance
- $b \in \mathbb{R}$, is a constant term that allows the activation function to be shifted in order to control the triggering threshold
- $v = w \times X + b$, weighted sum of entries
- $\forall$ $v \in \mathbb{R}$, 
$$
\hat{y} = 
\begin{cases}
1,& if & v \ge 0\\
0,& if & v < 0
\end{cases}
$$


## [▻]() $M_2^*$ | Parameter update using the **Stochastic Gradient Descent [SGD]** method

The update of the `weights` and the `bias` is done according to the [Stochastic Gradient Algorithm](https://en.wikipedia.org/wiki/Stochastic_gradient_descent) (the standard method of [gradient descent](https://en.wikipedia.org/wiki/Gradient_descent)) 

$$
w = w - 𝜂 \bigtriangledown Q(w)
$$
where, $𝜂$ is the iteration step size (sometimes called the learning rate in machine learning).

More precisely, according to Rosenblatt's rule, we have :
$$
w = w + 𝜂(y_j - \hat{y_j})\times X_j
$$
$$
b = b + 𝜂(y_j - \hat{y_j})
$$
where 
- $y_j$ the target value
- $\hat{y_j}$ the prediction(s) for each observation
- $y_j - \hat{y_j}$, the error
- $j$ -> each observation

***Training steps*** : the one used (several exist depending on the needs and complexity)*

> [BEGINNING]()
>
>1. Random initialization of weights and bias
>> $"loop"$ : 
> > {
> - For each observation $X_{j_{j \in \{{1,...,p}\}}}$ in $x_{i_{i \in \{{1,...,n}\}}}$, we have : 
>   $X_j = \{{x_{1j}, x_{2j}, ..., x_{nj}}\}$
> 2. Calculation of the weighted sum of the inputs (often denoted $z$) : 
>   $z = w \times X_j + b$
> 3. Application of the activation function (functions in [$M_1^*$]()) : 
>   $\hat{y} = 1$ si $z \ge 0$, $0$ sinon
> 4. Update of weights and bias (functions in [$M_2^*$]())
> > }
>
> [END]()


## [▻]() | Introductory summary

The [`perceptron`]() is a binary linear classifier. Its limitations (non-linearity, single layer, lack of regression) are not implementation errors but theoretical constraints. It remains useful as a basic model and as an introduction to neural networks. To overcome these limitations, [`MLPs`]() with multiple layers and differentiable activation functions are used.
